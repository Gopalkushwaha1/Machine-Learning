```python
# Find the number of movies staring with A for each group

def foo(group) :
  return group['Series_Title'].str.startswith('A').sum()

genres.apply(foo)
```

```python
# Find rating of each movies in the group according to IMDB score

def rank_movie(group) :
  group['genre_rank'] = group['IMDB_Rating'].rank(ascending=False)
  return group

genres.apply(rank_movie)
```

```python
# Find normlized IMDB rating group wise

def norm_rating(group) :
  group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()) / (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group
```

```python
genres.apply(norm_rating)
```
