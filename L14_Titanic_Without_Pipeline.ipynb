# Train Test Split

```python
X_train , X_test , y_train , y_test = train_test_split(df.drop(columns=['Survived']),df['Survived'],test_size=0.2,random_state = 42)
```

```python
X_train.head(2)
```

# Filling missing value

```python
si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy='most_frequent')

X_train_age = si_age.fit_transform(X_train[['Age']])
X_train_embarked = si_embarked.fit_transform(X_train[['Embarked']])

X_test_age = si_age.transform(X_test[['Age']])
X_test_embarked = si_embarked.transform(X_test[['Embarked']])
```

```python
X_train_age
```

```python
ohe_sex = OneHotEncoder( handle_unknown='ignore')
ohe_embarked = OneHotEncoder( handle_unknown='ignore')

X_train_sex = ohe_sex.fit_transform(X_train[['Sex']])
# Now apply OneHotEncoder directly to the Embarked column after filling missing values
X_train_embarked = ohe_embarked.fit_transform(X_train_embarked)

X_test_sex = ohe_sex.transform(X_test[['Sex']])
X_test_embarked = ohe_embarked.transform(X_test_embarked)
```

```python
X_train_sex.toarray()
```

## Scale numerical features

### Subtask:
Apply MinMaxScaler to the 'Age' and 'Fare' columns.


**Reasoning**:
Apply MinMaxScaler to the 'Age' and 'Fare' columns in both the training and testing sets.



```python
scaler = MinMaxScaler()

X_train_scaled_age = scaler.fit_transform(X_train[['Age']])
X_test_scaled_age = scaler.transform(X_test[['Age']])

X_train_scaled_fare = scaler.fit_transform(X_train[['Fare']])
X_test_scaled_fare = scaler.transform(X_test[['Fare']])
```

```python
X_train_rem = X_train.drop(columns=['Sex','Age','Embarked'])
X_test_rem = X_test.drop(columns=['Sex','Age','Embarked'])
#
```

```python
X_train_transformed = np.concatenate((X_train_rem, X_train_age, X_train_sex.toarray(), X_train_embarked.toarray()), axis=1)
X_test_transformed = np.concatenate((X_test_rem, X_test_age, X_test_sex.toarray(), X_test_embarked.toarray()), axis=1)
```

```python
X_test_transformed.shape
```

```python
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train_transformed, y_train)
```

```python
y_pred = clf.predict(X_test_transformed)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
