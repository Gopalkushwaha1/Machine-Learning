```python
# Label title

plt.title('Rohit vs Virat ')
plt.xlabel('Year')
plt.ylabel('Runs Score ')
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# Color for graph

plt.title('Rohit vs Virat ')
plt.xlabel('Year')
plt.ylabel('Runs Score ')
plt.plot(batsman['index'],batsman['V Kohli'],'g')
plt.plot(batsman['index'],batsman['RG Sharma'],'b')
```

```python
# Change line


plt.title('Rohit vs Virat ')
plt.xlabel('Year')
plt.ylabel('Runs Score ')
plt.plot(batsman['index'],batsman['V Kohli'],'g' , linestyle='dotted')
plt.plot(batsman['index'],batsman['RG Sharma'],'b')
```

```python
# Denote Point

plt.title('Rohit vs Virat ')
plt.xlabel('Year')
plt.ylabel('Runs Score ')
plt.plot(batsman['index'],batsman['V Kohli'],'g' , linestyle='dotted' , marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],'b')
```

```python
# Name the line name


plt.title('Rohit vs Virat ')
plt.xlabel('Year')
plt.ylabel('Runs Score ')
plt.plot(batsman['index'],batsman['V Kohli'],'g' , linestyle='dotted' , marker='D' , label='Kohli')
plt.plot(batsman['index'],batsman['RG Sharma'],'b' , label = 'Rohit')

plt.legend(loc='best')
```

```python
# grid

# Name the line name


plt.title('Rohit vs Virat ')
plt.xlabel('Year')
plt.ylabel('Runs Score ')
plt.plot(batsman['index'],batsman['V Kohli'],'g' , linestyle='dotted' , marker='D' , label='Kohli')
plt.plot(batsman['index'],batsman['RG Sharma'],'b' , label = 'Rohit')

plt.legend(loc='best')
plt.grid()
```

# Bar Plot

```python
# Simple bar chat

childern = [10,14,9]
color = ['red','green','pink']

plt.bar(color,childern , color = 'green')
```

```python
# Simple bar chat horijental

childern = [10,14,9]
color = ['red','green','pink']

plt.barh(color,childern , color = 'green')
```

```python
df = pd.read_csv('/content/batsman_season_record.csv')
```

```python
df
```

```python
plt.bar(df['batsman'],df['2015'])
```

```python
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'])
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2 ,df['2015'] , width=0.2)
plt.bar(np.arange(df.shape[0]) ,df['2016'] , width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2 ,df['2017'] , width=0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
```

```python
# Problem suppose color name is to large

childern = [10,14,9]
color = ['redrrrrrrrrrrrrrrrrrrrrrrreeeeeeeeeeeeeeddddddddddddddddwwwwwwww','green','pink']

plt.bar(color,childern , color = 'green')
```

```python
# to solve prev problem big name we use xtrick function

childern = [10,14,9]
color = ['red','green','pink']

plt.barh(color,childern , color = 'green')
plt.xticks(rotation = 'vertical')
```

```python
# Stacked bar chart

plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],label='2016',bottom=df['2015'])

plt.legend()
```
