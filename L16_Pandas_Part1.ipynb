# Series from lists

```python
# String

country = ['India','Russia','USA','China','Japan']

pd.Series(country)
```

```python
# Integer

num = [1,2,3,4,5]

pd.Series(num)
```

```python
# Custom Index

num = [1,2,3,4,5]

pd.Series(num,index = ['a','b','c','d','e'])
```

```python
from os import name
# Setting a name

num = [1,2,3,4,5]

pd.Series(num,index=['a','b','c','d','e'],name='Sample')
```

```python
# Series from dict

marks = {
    'Hindi':82 ,
    'Math' : 90 ,
    'Phy' : 35 ,
    'Chem' : 70
}

m = pd.Series(marks , name='Class 12th Marks')
```

```python
m
```

# Series Methods

```python
# Head    -> Top 5 , we also see custom as top

name_series.head()
```

```python
# Tail    -> last 5 , we also see custom as last

name_series.head()
```

```python
# Custom

n = name_series

n.head(10)
```

```python
# Random

n.sample()
```

```python
# Custom random

n.sample(10)
```

```python
# Values count -> Count freq

n.value_counts()
```

```python
# Sort Values -> inplace

n.sort_values()
```

```python
# Sort Index -> Inplace it do permanent change

n.sort_index(inplace = True)
```
