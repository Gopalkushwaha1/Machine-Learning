```python
# Relational Plot

# Use to see statistical relation between 2 or more variables.
# Bivariate
```

```python
# Scatter plot --> relational plot type

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.sample()
```

```python
# Scatter plot  - it is axix level function

sns.scatterplot(tips,x='total_bill',y='tip')
```

# Figure level function

```python
sns.relplot(tips,x='total_bill',y='tip')    # Same graph as above but it use rel obj while upeer one use axix object
```

```python
sns.relplot(tips,x='total_bill',y='tip',kind='scatter')
```

```python
sns.relplot(tips,x='total_bill',y='tip',hue='sex')
```

```python
sns.relplot(tips,x='total_bill',y='tip',hue='sex',style='time')
```

```python
sns.relplot(tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

# Distribution Plots

* Used for univariate analysis

* Are the outliers.?

* Is data bimodel

```python
# Histplot   --> Distribuation plot type

sns.histplot(tips['total_bill'])
```

```python
# Figure level -> displot
# Axex level  -> Histplot , kdplot , rugplot
```

```python
# Plotting univariate histogram

sns.displot(tips['total_bill'])
```

```python
df = sns.load_dataset('titanic')
```

```python
# Kdeplot -> 2nd Type
# Kernal density
```

# Matrix plot


```python
gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
temp_df =  gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(temp_df)
```

```python
temp = gap[gap['continent'] == 'Asia'].pivot(index='country',columns='year',values='lifeExp')

sns.heatmap(temp , annot=True,linewidths=0.5,cmap='Blues')
```
