```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
iris = sns.load_dataset('iris')
```

```python
sns.scatterplot(x='total_bill', y='tip', data=tips)
```

```python
# Strip plot
# It is axix level function
sns.stripplot(x='day', y='total_bill', data=tips)
```

```python
# Using catplot
# It is figure level function
sns.catplot(x='day', y='total_bill', data=tips)
```

```python
# jitter

sns.catplot(x='day', y='total_bill', data=tips , jitter=False)
```

```python
# jitter

sns.catplot(x='day', y='total_bill', data=tips , jitter=0.2)
```

```python
# hue

# jitter

sns.catplot(x='day', y='total_bill', data=tips , jitter=False , hue = 'sex')
```

```python
# Swamplot

sns.catplot(x='day', y='total_bill', data=tips , kind = 'swarm')
```

```python
sns.pairplot(data=iris , hue = 'species')
```

# Pair Grid

```python
sns.PairGrid(data=iris)
```

```python
g = sns.PairGrid(data=iris)
g.map(sns.scatterplot )
```

```python
# map_diag -> map_offdiag

g = sns.PairGrid(data=iris , hue = 'species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
# map_diag -> map_upper -> map_lower

g = sns.PairGrid(data=iris , hue = 'species')
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
```

```python
# vars -> To work only spefic col

g = sns.PairGrid(data=iris , hue = 'species' , vars = ['sepal_length','sepal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
```
